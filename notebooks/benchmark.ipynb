Reproduces the 1,000-pair synthetic benchmark from the README and charts the verdicts. Runs in Google Colab.

```python
import csv, collections
import matplotlib.pyplot as plt
from naam_match import match
rows = list(csv.DictReader(open('benchmark/pairs.csv', encoding='utf-8')))
kinds = collections.defaultdict(collections.Counter)
for r in rows:
    kinds[r['kind']][match(r['name_a'], r['name_b']).verdict] += 1
labels = sorted(kinds)
fig, ax = plt.subplots(figsize=(10, 5))
left = [0] * len(labels)
for v, c in [('Match', '#3B5BDB'), ('Send for review', '#D9622B'), ('No match', '#8E8E93')]:
    vals = [kinds[k][v] for k in labels]
    ax.barh(labels, vals, left=left, label=v, color=c)
    left = [a + b for a, b in zip(left, vals)]
ax.set_xlabel('pairs'); ax.set_title('naam-match verdicts by variation type (synthetic benchmark)'); ax.legend(); plt.tight_layout(); plt.show()
```
